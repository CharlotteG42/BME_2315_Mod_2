## Project Goal:

FSHD is an inherited muscle disease caused by abnormal expression of the DUX4 gene in skeletal muscle, and it progresses slowly and unevenly as muscle is replaced by fat (Wong et al., 2024). Because strength, walking, and stair tests and clinical severity scores change little from year to year, a trial that relies on them would need many participants and a long follow-up, so an objective MRI measure such as fat fraction (the share of the MRI signal that comes from fat) could serve as an exploratory endpoint. Using natural-history data from 34 people with FSHD, we compare tibialis anterior fat fraction (left, right, and bilateral mean), fat-fraction variability, dorsiflexion strength, timed up-and-go, 10 m walk/run, stair times, and the clinical severity score, linearly interpolating each person's values to 11 and 22 months and summarizing sensitivity to change with the standardized response mean (mean change divided by the standard deviation of the change). Our primary question is: do people with a larger initial fat fraction show a greater change in fat fraction over time? Our supporting question is: which measure is most sensitive to change over 11 and 22 months? These questions matter because, if starting level predicts change, a trial can enroll people in the range where change is largest and need fewer participants, and the comparison shows which measures are worth including as exploratory endpoints. 

 

## Disease Background:

### Prevalence & incidence
Facioscapulohumeral muscular dystrophy (FSHD) is one of the more common muscular dystrophies in adults, with an estimated prevalence of about 4-10 cases per 100,000 people (Preston M. and Wang L., 2025), or roughly 1 case per 8,000 individuals (Muscular Dystrophy Association, 2026). A comprehensive, population-based study in the Netherlands found a prevalence of about 12 per 100,000 people, implying that many cases exist with less noticeable symptoms (Deenen J. et al., 2014). The same study reported an incidence of 0.3 per 100,000 person-years, or about 52 new diagnoses per year (Deenen J. et al., 2014). For a disease as rare as FSHD, this is a rather large patient population, which makes a clinical trial more realistic, but it also means trials have to account for a wide range of severity.

### Symptoms
FSHD is named after the muscles it tends to affect first: the face (trouble closing the eyes, smiling, or whistling), the muscles around the shoulder blades (scapular winging, the most common first sign), and the upper arms (FSHD Society, 2019). Weakness is often asymmetric, meaning one side can be much weaker than the other. Many people also develop weakness in the lower legs, starting with foot drop from a weak tibialis anterior, which is the muscle in our MRI data. However, weakness is not the only symptom: about 82% of patients report chronic pain, fatigue is the most common complaint after weakness, and some have mild hearing loss or heart rhythm changes (Preston M. and Wang L., 2025). Symptoms usually start in teenage years, but onset varies, and some people with the genetic change live their full lives without ever noticing symptoms. Since the TA controls lifting the foot, weakness there connects directly to the walking and stair tests in our dataset.

### Diagnosis
FSHD is diagnosed with genetic testing and clinical evaluation. Blood testing identifies about 95% of cases, known as FSHD1, where a repeated stretch of DNA on chromosome 4 called D4Z4 is shortened to 1-10 repeat units (normal is 11 or more) on a "permissive" version of the chromosome called 4qA. The other 5% of cases are FSHD2, where the repeat length is normal but the region loses its methylation, most often because of a variant in the SMCHD1 gene. Testing measures the D4Z4 repeat size, checks the chromosome version and methylation, and sequences the FSHD2 genes. Blood tests like creatine kinase (CK) are usually normal to only a little high, and muscle biopsies don't show anything specific to FSHD. FSHD1 is autosomal dominant, so a parent with it has a 50% chance of passing it on, but 10 to 30% of cases are new mutations with no family history (University of Rochester Medical Center, n.d.). Our dataset includes each subject's D4Z4 repeat count, which could help explain differences in severity.

### Standard of care treatments (& reimbursement)
There is currently no approved treatment that slows or stops FSHD, so care focuses on managing symptoms (Voermans N., 2023). This includes physical therapy and moderate aerobic exercise, ankle-foot braces for foot drop, surgery to fix the shoulder blade to the ribcage in some patients to improve arm range of motion, pain management, breathing support at night when needed, and eye drops for people who can't fully close their eyes. The disease is also expensive: a Dutch study of 172 adults with FSHD estimated a total cost of about €26,322 per patient per year, including €12,077 in healthcare, €9,179 in non-medical costs like care and equipment, and €5,066 in lost work productivity (Blokhuis A. et al., 2021). When the effect on quality of life was included, the total rose to about €40,850. With no real treatment and costs this high, there is great need for trials that can show whether a new therapy works.

### Disease progression & prognosis
FSHD usually gets worse slowly over decades. About 20% of patients eventually need a wheelchair, but life expectancy is generally normal, as severe cardiac involvement or breathing failure is rare (Preston M. and Wang L., 2025). Progression also happens muscle by muscle rather than all at one time. In a one-year MRI study, the overall fat in the legs barely changed, but individual muscles changed drastically, and muscles that started at a medium fat level (40 to 50%) progressed most rapidly (Wang L. et al., 2021). In that study, the tibialis anterior changed very little over one year, and TA strength dropped sharply once more than about 20% of the muscle was replaced by fat. Muscles that are bright on STIR MRI (a sign of inflammation or edema) are much more likely to get fattier: 13.9% of STIR-positive muscles had more fat after a year, compared to only 0.21% of STIR-negative muscles (Monforte M. et al., 2019). Meanwhile, functional tests like the 6-minute walk didn't change noticeably in a year. This is the core problem for our project: a short trial needs a measure that changes within months, and MRI can, but only in certain muscles, while function may lag behind.

### Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
FSHD is caused by a gene called DUX4 that is inside each D4Z4 repeat. DUX4 is normally switched off in adult muscle, but when the repeat array is shortened (FSHD1) or loses its methylation (FSHD2), the DNA opens. On the 4qA chromosome version, the DUX4 message gets a signal that makes it stable, so muscle cells start making DUX4 protein. DUX4 turns on genes that are normally only active in early embryos, along with immune-related genes, and this is toxic to muscle cells and leads to cell death (Geng L. et al., 2012). At the tissue level, the lost muscle fibers are replaced by fat and scar tissue, along with chronic inflammation. Fat replacement is what the MRI fat fraction measures in our data set. In the study our dataset comes from, TA muscles that were STIR-positive had more fat and higher DUX4 activity, every TA over 20% fat was STIR-positive, and higher DUX4 activity was linked to lower TA strength and worse clinical severity (Wong C. et al., 2024). At the whole muscle level, strength depends on both how much muscle there is and its quality, since fat infiltration lowers how much force each part of the muscle can produce. This explains why fat fraction should be related to strength.

### Clinical trials/next-gen therapies
Since DUX4 is the root cause of FSHD, most new therapies try to turn it off. One attempt, losmapimod, made it to a Phase 3 trial but didn't beat placebo on its main measure, reachable workspace, and was stopped in 2024 (Kinoshita J., 2024). Delpacibart braxlosiran (del-brax), owned by Novartis, uses an antibody attached to a small piece of RNA to silence DUX4 directly in muscle. In a Phase 1/2 trial, it lowered the blood levels of a DUX4-regulated biomarker and of creatine kinase, a marker of muscle damage, with a safety profile consistent with earlier results. It is now enrolling a 200-patient Phase 3 trial that measures muscle strength in the US and walking speed in Europe (Novartis, 2026). EPI-321, made by Epicrispr Biotechnologies, takes a different approach: it's a one-time IV infusion that uses epigenetic editing to keep DUX4 switched off without changing the DNA itself. Early Phase 1/2 results showed patients gained from 0.5 to 1.3 pounds of lean muscle on MRI after six months with no serious side effects so far, though only three patients had reached that point (Horak M., 2026).

## MRI Data Analysis Methodology:
For this section, we reproduced how the fat fraction values in `Module2_FSHD_Dataset.csv` are calculated from raw MRI images. We used one example subject (UW_0020) and looked at the left tibialis anterior (TA) at two visits (TP1 and TP2), with 8 locations along the muscle at each visit.

We followed the code from the class example notebook and moved its functions into `fshd.py` (loading, checking, calculating, and summarizing the data) and `plots.py` (the figures), so `main.py` can also run the whole analysis on its own. We used `numpy` for the image math, `pandas` for tables, and `matplotlib` for the figures.

Compared to the class example, we added data checks, a leg-tissue comparison, and a grid of fat fraction maps (Figure 4). We also fixed the example's profile plot, which labeled the x-axis in mm but actually plotted percent, and saved both visits to the CSV instead of just TP1.

> To run this notebook, keep `fshd.py`, `plots.py`, and the `FSHD_TA_MRI_data_0020_TP1_TP2` folder in the same folder as this notebook.

```python
from fshd import *
import plots
import numpy as np
import pandas as pd
```

### 1. Loading the MRI Data
`slice_info.csv` tells us which water image, fat image, and mask belong together. There are 16 rows: 8 locations along the muscle at each of the 2 visits. Each row also gives the location as a percent of the muscle's length and as a distance (mm) from the start of the muscle.

```python
manifest = load_manifest()
print(f"Number of samples = {len(manifest)}")      # should be 16

# Distance from the start of the muscle (mm) for each location and visit
manifest.pivot(index="percent_along_muscle", columns="timepoint",
               values="distance_from_muscle_start_mm")
```

> **Note:** Both visits were sampled at the same percent of the muscle's length, but the distance in mm doesn't always match (it gets up to 12 mm apart at 60-80%). So when we compare visits, we're comparing the same relative spot, not the exact same slice of tissue.

### 2. Checking the Data for Errors
Before calculating anything, we ran six checks on every sample using `check_sample()`: the three images are the same size, the mask only contains 0s and 1s, there are no missing values, no signal is negative, every TA pixel has signal (so we never divide by zero), and the mask isn't empty. All 16 samples passed.

```python
# PSEUDOCODE:
#   start a counter at 0
#   for each of the 16 samples (each row of the manifest):
#       run the 6 checks on its water, fat, and mask files
#       if every check is True, add 1 to the counter
#       if any check is False, print which sample failed
#   print the names of the checks and how many samples passed

samples_passed = 0

for _, row in manifest.iterrows():
    checks = check_sample(row)
    if all(checks.values()):        # True only if every check passed
        samples_passed += 1
    else:
        print(f"Failed a check: {row['timepoint']} slice {row['slice_id']}")

print(f"Checks run on each sample: {list(checks.keys())}")
print(f"Samples that passed every check = {samples_passed} of "
      f"{len(manifest)}")
```

### 3. Calculating Fat Fraction for One Sample
We started with one mid-muscle sample (TP1, 40% along the muscle), the same one used in the class example. For each pixel, fat fraction is:

**FF (%) = 100 × fat / (water + fat)**

We only calculated it where water + fat is above 0 so we never divide by zero, and set pixels outside the TA mask to `NaN` so they're left out of the TA results. To see why the mask matters, we compared three regions: the whole supplied image, leg tissue only (water + fat above 20), and the TA only.

```python
# --- 3a: Loading one sample ---
# Representative mid-muscle slice (TP1, 40% along the muscle)
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]
water, fat, mask = load_sample(row)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())

denominator = water + fat
median_signal = np.median(denominator[denominator > 0])
print(f"Median signal of pixels with any signal = {median_signal:.0f}")

# --- 3b: Comparing the three regions ---
whole_image_values, leg_values, muscle_values = region_values(water, fat,
                                                              mask)

region_rows = []
for region_name, values in [("Whole supplied region", whole_image_values),
                            ("Leg tissue only", leg_values),
                            ("TA muscle only", muscle_values)]:
    region_rows.append({
        "Region": region_name,
        "Pixels": len(values),
        "Mean FF (%)": np.mean(values),
        "Median FF (%)": np.median(values),
        "Min (%)": np.min(values),
        "Max (%)": np.max(values),
    })

pd.DataFrame(region_rows).round(1)
```

> **Note:** About 3 out of 4 pixels in the whole image aren't tissue at all; they're background noise outside the leg with tiny signals (median of 3, compared to hundreds in tissue). With numbers that small, FF lands on values like 0%, 50%, or 100% (fat = 1 and water = 1 gives 50%), so the whole-image mean of 48.6% is mostly noise from the MRI. Leg tissue mixes fat and bone marrow with muscle, so its mean doesn't describe any one tissue either. Only the TA value (29.9%) tells us about this muscle.

Next, we plotted histograms of fat fraction for the three regions. Since the regions have very different numbers of pixels, we used density (fraction of pixels per bin) instead of counts so they can be compared on the same graph.

```python
plots.plot_histogram(whole_image_values, leg_values, muscle_values,
                     "Fat-fraction distributions, TP1 at 40% along the TA")
```

The TA is right-skewed: most pixels are lean, but a tail of fatty pixels pulls the mean (29.9%) above the median (25.4%), which is why we report both.

### 6. Summarizing Every Sample at Both Visits
We then used `summarize_visit()` to repeat the calculation for every location at TP1 and TP2. For each sample, it records the mean, median, and standard deviation of TA fat fraction and the number of TA pixels. The two visits are then combined into one table with `pd.concat`. All fat fraction values are in %.

```python
# PSEUDOCODE:
#   for each visit (TP1, then TP2):
#       for each of the 8 locations along the muscle:
#           load water, fat, and mask, then calculate TA fat fraction
#           record the mean, median, SD, and number of TA pixels
#   stack the TP1 and TP2 tables into one table (pd.concat)
#   show it with shorter column names so it fits on the page

profiles_df = pd.concat(
    [summarize_visit(manifest, "TP1"), summarize_visit(manifest, "TP2")],
    ignore_index=True
)

# Shorter column names so the table fits on the page
short_names = {
    "timepoint": "visit",
    "percent_along_muscle": "% along",
    "distance_from_muscle_start_mm": "distance (mm)",
    "mean_fat_fraction_percent": "mean FF",
    "median_fat_fraction_percent": "median FF",
    "std_fat_fraction_percent": "SD",
    "n_muscle_pixels": "TA pixels",
}
profiles_df[list(short_names)].rename(columns=short_names).round(1)
```

### 7. Fat Fraction Along the Muscle - Figure 3
To see how fat fraction changes along the muscle, we plotted the mean and median at each location for both visits. We put distance (mm) on the x-axis instead of percent, since the samples aren't the same distance apart at each visit.

```python
plots.plot_profile(profiles_df)
```

**Figure 3.** Mean (solid, ± SD) and median (dashed) TA fat fraction along the muscle for TP1 (blue) and TP2 (orange). Fat fraction is highest near the start of the muscle (about 80-85%) and lowest in the middle (about 17% at TP1). At TP2, the fatty region reaches further down the muscle, and every location from 30% on is higher. This means a single slice could give very different results depending on where it was taken.

> **Note:** The first two locations only have 46 to 115 TA pixels because the muscle tapers there, so a small error in the mask has a bigger effect on their values.

### 8. Comparing TP1 vs. TP2
To compare the visits, `compare_visits()` makes a table with one row per location and one column per visit, then adds a column for the change (TP2 - TP1). We did this for mean fat fraction and for the number of TA pixels.

```python
# PSEUDOCODE:
#   make a table with one row per location and one column per visit
#       (TP1, TP2) for the mean fat fraction
#   add a column for the change: TP2 - TP1
#   do the same for the number of TA pixels
#   count how many locations had a higher mean at TP2

mean_comparison = compare_visits(profiles_df)
pixel_comparison = compare_visits(profiles_df, "n_muscle_pixels")

# Counting how many locations had a higher mean fat fraction at TP2
locations_increased = 0
for change in mean_comparison["change_TP2_minus_TP1"]:
    if change > 0:
        locations_increased += 1

print(f"Mean FF increased at {locations_increased} of "
      f"{len(mean_comparison)} locations")
print("\nMean fat fraction (%):")
display(mean_comparison.round(1))
print("Number of TA pixels:")
display(pixel_comparison)
```

> **Note:** From 30% on, fat fraction went up by 8 to 26 points, and the number of TA pixels went down at every location. Both fit with muscle being replaced by fat and shrinking, but the pixel count could also change if the mask was drawn differently at TP2. We also don't know how much time passed between visits, so this is a change between visits, not a yearly rate. As the lectures pointed out, a change in an MRI biomarker doesn't automatically mean the patient's function got worse.

```python
# PSEUDOCODE:
#   for each of the 8 locations along the muscle:
#       load the TP1 and TP2 samples and calculate TA fat fraction
#       find one box that fits the TA at both visits
#           (so both visits are zoomed in the same amount)
#       draw the TP1 map with the TP2 map right below it
#   use the same 0-100% color scale for every map

plots.plot_map_grid(manifest)
```

### 11. Assumptions and Next Steps
Our analysis relies on some assumptions:
- Fat fraction is calculated from the signal with no correction for other MRI effects, so it's an estimate of true fat content.
- The TA masks are accurate. Errors matter most at the ends, where the muscle is small.
- Each value comes from one 4 mm slice, not the whole muscle volume.
- Visits were matched by percent of muscle length, not exact position.

### Part A: Interpolating Along the Muscle

### 1. The Sampled Locations
Our MRI analysis gave the mean TA fat fraction at 8 locations along the muscle at each visit (`profiles_df`, MRI section 6). These are the sampled points we interpolate between. We used **percent of muscle length** as the location variable instead of mm, because the mm distances don't match between visits (MRI section 1).

```python
# Sampled points: mean TA fat fraction (%) at each location and visit
sampled = profiles_df.pivot(index="percent_along_muscle",
                            columns="timepoint",
                            values="mean_fat_fraction_percent")
sampled.round(1)
```

### 2. Estimating Fat Fraction at 45% - Figure 5
We chose **45%** because no image was taken there, and it's halfway between the 40% and 50% slices, where fat fraction changed the most between visits (MRI section 8). `interpolate_location()` draws straight lines between the sampled locations with `interp1d`, the same function as the class example, and returns the value at 45%. It also checks that the target is inside the sampled range (10% to 80%); a location outside it would be extrapolation, so it returns `NaN`.

```python
# PSEUDOCODE:
#   for each visit (TP1, TP2):
#       linear estimate of fat fraction at 45% (between 40% and 50%)
#   print both estimates and the change between visits
#   check: a location outside 10-80% returns NaN (extrapolation)

TARGET_PERCENT = 45
estimates = {}
for timepoint in ["TP1", "TP2"]:
    estimates[timepoint] = interpolate_location(profiles_df, timepoint,
                                                TARGET_PERCENT)
    print(f"{timepoint}: estimated fat fraction at {TARGET_PERCENT}% = "
          f"{estimates[timepoint]:.1f}%")

change_45 = estimates["TP2"] - estimates["TP1"]
print(f"Change between visits at {TARGET_PERCENT}% = "
      f"{change_45:.1f} points")

# 90% is past the last slice (80%), so this should be NaN
print(f"\nEstimate at 90% (outside the sampled range) = "
      f"{interpolate_location(profiles_df, 'TP1', 90)}")
```

```python
plots.plot_location_interpolation(profiles_df, TARGET_PERCENT, estimates)
```

**Figure 5.** Mean TA fat fraction at the 8 imaged locations (dots) and the linear estimates at 45% (diamonds) for TP1 (blue) and TP2 (orange). Each estimate sits on the straight line between the 40% and 50% slices: 25.4% at TP1 and 46.3% at TP2, an increase of 20.9 points. Because 45% is between two measured slices, these are interpolations, not extrapolations.

### 3. Checking the Estimate Against Real Slices
An interpolated value is an estimate, not a measurement, so we tested how close it gets to a real one. For each inside location (20% to 70%), we pretended that slice was never imaged, estimated it from its two neighbors with `leave_one_out()`, and compared the estimate to the real value.

```python
# PSEUDOCODE:
#   for each visit and each inside location (20% to 70%):
#       estimate it from the two neighboring slices (straight line)
#       error = estimate - real value
#   print the average and largest error (ignoring + or -)

check_table = pd.concat([leave_one_out(profiles_df, "TP1"),
                         leave_one_out(profiles_df, "TP2")],
                        ignore_index=True)
print(f"Average error = {check_table['error'].abs().mean():.1f} points")
print(f"Largest error = {check_table['error'].abs().max():.1f} points")
check_table.round(1)
```

> **Note:** Where fat fraction is fairly flat (50-60%), the estimates were within 5 points of the real slice, but where it drops quickly (20-40%), they missed by up to 12 points. These tests used neighbors 20% apart; our 45% estimate uses slices only 10% apart, so it should be closer. It still sits on the steepest part of the curve, so it could be off by several points, which is why an interpolated value can't replace actually imaging that location.

### 9. Comparing Sensitivity to Change (SRM) - Figure 8
To compare measures with different units (%, kg, seconds, points), we used the **standardized response mean (SRM)**: the mean change divided by the standard deviation of the change. It is the same mean ÷ SD idea from Lecture 4. A larger SRM means the measure changes in the same direction for most subjects, which is what a trial needs to detect an effect. Common cutoffs are about 0.2 (small), 0.5 (moderate), and 0.8 (large).

Since strength goes *down* as the disease worsens, we flipped its sign so that **positive SRM always means getting worse**. We also used a **paired t-test** (baseline vs. interpolated value for the same subjects) to test whether each change is significantly different from zero.

```python
# PSEUDOCODE:
#   for each measure, at T1 and T2:
#       SRM = mean change / SD of change (positive = worse)
#       % worse = percent of subjects who got worse
#       paired t-test: baseline vs. estimate at T
#   show the table and the bar chart

summary = summarize_sensitivity(results)

# Shorter column names so the table fits on the page
short_names = {
    "label": "measure",
    "n_T2": "n (T2)",
    "mean_change_T2": "mean change T2",
    "sd_change_T2": "SD change T2",
    "SRM_T1": "SRM T1",
    "SRM_T2": "SRM T2",
    "pct_worse_T2": "% worse T2",
    "p_T2": "p (T2)",
}
summary_table = summary[list(short_names)].rename(columns=short_names)
summary_table = summary_table.sort_values("SRM T2", ascending=False)
summary_table.round(3)
```

```python
plots.plot_srm(summary)
```

**Figure 8.** Standardized response mean for each measure from baseline to T1 (about 11 months, blue) and T2 (about 22 months, orange); positive means the measure got worse. Only the three TA fat fraction measures move clearly outside the "trivial" gray zone, and the average of both legs has the largest SRM (0.69 at T1, 0.78 at T2, a moderate-to-large effect). It got worse in 88% of subjects by T2 and was the only kind of measure with a significant paired t-test (p < 0.001). Every strength, functional, and clinical measure stayed near zero (SRM between -0.16 and 0.26, all p > 0.19).

> **Note:** We ran 24 t-tests (12 measures x 2 time points), so a few could be "significant" by chance. A strict Bonferroni correction (0.05 / 24 = 0.002) still keeps the both-legs fat fraction significant at both time points (p = 0.0007 and 0.0009). Averaging both legs gave a higher SRM than either leg alone because it reduces random measurement noise.

### 10. Primary Question: Does Baseline Fat Fraction Predict Change? - Figure 9
To answer our primary question, we plotted each subject's baseline fat fraction (both legs) against their change by T2. We used two correlation tests:
- **Pearson r** tests for a straight-line relationship (the same test we used in Module 1).
- **Spearman rho** tests whether change tends to rise as baseline rises, without assuming a straight line, and is less affected by a few extreme points.

We also split subjects at **20% baseline fat fraction**, the point where Wang et al. (2021) found TA strength drops steeply and Wong et al. (2024) found every muscle was STIR-positive, and compared the two groups with **Welch's t-test** (which doesn't assume equal variance, since the groups are very different sizes).

```python
# PSEUDOCODE:
#   baseline fat fraction vs. change by T2 (both legs)
#   Pearson r and Spearman rho (with p-values)
#   split at 20% baseline fat fraction and compare with Welch's t-test

correlation = baseline_vs_change(results, "ff_mean", "T2")
print(f"n = {correlation['n']}")
print(f"Pearson r = {correlation['pearson_r']:.2f}, "
      f"p = {correlation['pearson_p']:.3f}")
print(f"Spearman rho = {correlation['spearman_rho']:.2f}, "
      f"p = {correlation['spearman_p']:.3f}")

groups = compare_groups(results, "ff_mean", "T2", threshold=20)
print(f"\nBelow 20%: n = {len(groups['low'])}, "
      f"mean change = {groups['low'].mean():.2f} points")
print(f"20% or above: n = {len(groups['high'])}, "
      f"mean change = {groups['high'].mean():.2f} points")
print(f"Welch's t-test: t = {groups['t_stat']:.2f}, "
      f"p = {groups['p_value']:.3f}")
```

```python
plots.plot_baseline_vs_change(correlation, threshold=20)
```

**Figure 9.** Baseline TA fat fraction (both legs) vs. change in fat fraction by T2 for 24 subjects, with a linear regression line. The Spearman correlation is moderate and significant (rho = 0.52, p = 0.010), but the Pearson correlation is weak and not significant (r = 0.25, p = 0.231). The difference comes from the shape of the data: most subjects with very little fat (9 of 11 under 10%) changed by less than 1.5 points, subjects between about 10% and 65% changed the most (up to 7.3 points), and the subject with the most fat (76%) actually decreased. This "rise then fall" pattern is a ceiling effect: a muscle that is already mostly fat has little muscle left to replace. Subjects at or above 20% changed about twice as much on average (3.3 vs. 1.6 points), but with only 5 subjects in that group the difference was not significant (p = 0.365).

So the answer to our primary question is **"yes, up to a point."** Higher baseline fat fraction goes with larger change, except in muscles that are already mostly fat.

### 12. Interpolation Assumptions and Limitations
- **Interpolation is an estimate, not a measurement.** The value at 45% and the values at T1 and T2 are what the measure *probably* was, based on the real measurements around them.
- **Straight lines between slices (Part A).** We assumed fat fraction changes evenly between neighboring slices. The slices are 10% of the muscle apart, and where fat fraction changes quickly the estimate can be off by up to about 12 points (section 3). Part A also uses only one subject.
- **Linear change between visits (Part B).** We assumed each measure changed at a steady rate between visits. The quadratic check (section 11) showed this choice made little difference for fat fraction (0.24 points on average).
- **Within the measured range.** 45% is between two imaged slices, and T1 and T2 are the earliest Year 1 and Year 2 visits, so every estimate is within the range of the measured data.
- **Missing data.** Only 21-25 subjects had all three visits for each measure, which limits statistical power, especially for the 20% group comparison (5 subjects).
- **Noisy functional tests.** Effort, mood, and day-to-day variation affect strength and timed tests, so a single subject's interpolated value may reflect noise rather than real change (Figure 6).
- **Fat fraction variability direction.** We counted an increase in fat fraction variability as "worse," but it is less clear what direction is meaningful for this measure.
