## Disease Background: 

### Prevalence & incidence
Facioscapulohumeral muscular dystrophy (FSHD) is one of the more common muscular dystrophies in adults, with an estimated prevalence of about 4-12 cases per 100,000 people (roughly 1 case per 8.000 individuals) [1]. A comprehensive, population-based study in the Netherlands found a prevalence of about 12 per 100,000 people, implying that many cases exist with less noticeable symptomes. The same study reported an incidence of 0.3 per 100,000 person-years, or about 52 new diagnoses per year [2]. For a disease as rare as FSHD, this is a rather large patient population, which makes a clinical trial more realistic, but it also means trials have to account for a wide range of severity.

### Symptoms
FSHD is named after the muscles it tends to affect first: the face (trouble closing the eyes, smiling, or whistling), the muscles around the shoulder blades (scapular winging, the most common first sign), and the upper arms [3]. Weakness is often asymmetric, meaning one side can be much weaker than the other. Many people also develop weakness in the lower legs, starting with foot drop from a weak tibialis anterior, which is the muscle in our MRI data. However, weakness is not the only symptom: about 82% of patients report chronic pain, fatigue is the most common complaint after weakness, and some have mild hearing loss or heart rhythm changes [4]. Symptoms usually start in teeange years, but onset varies, and some people with the genetic change live their full lives without ever noticing symptoms. Since the TA controls lifting the foot, weakness there connects directly to the walking and stair tests in our dataset.

### Diagnosis
FSHD is diagnosed with genetic testing and clinical evaluation. Blood testing indetifies about 95% of cases, known as FSHD1, where a repeated stretch of DNA on chromosome 4 called D4Z4 is shortened to 1-10 repeat units (normal is 11 or more) on a “permissive” version of the chromosome called 4qA. The other 5% of cases are FSHD2, where the repeat length is normal but the region loses its methylation, most often because of a variant in the SMCHD1 gene. Testing measures the D4Z4 repeat size, checks the chromosome version and methylation, and sequences the FSHD2 genes. Blood tests like creatine kinase (CK) are usually normal to only a little high, and muscle biopsies don't show anything specific to FSHD. FSHD1 is autosomal dominant, so a parent with it has a 50% chance of passing it on, but 10 to 30% of cases are new mutations with no family history [5]. Our dataset includes each subject's D4Z4 repeat count, which could help explain differences in severity.


### Standard of care treatments (& reimbursement)
There is currently no approved treatment that slows or stops FSHD, so care focuses on managing symptoms [6]. This includes physical therapy and moderate aerobic exercise, ankle-foot braces for foot drop, surgery to fix the shoulder blade to the ribcage in some patients to improve arm range of motion, pain management, breathing support at night when needed, and eye drops for people who can't fully close their eyes. The disease is also expensive: a Dutch study of 172 adults with FSHD estimated a total cost of about €26,322 per patient per year, including €12,077 in healthcare, €9,179 in non-medical costs like care and equipment, and €5,066 in lost work productivity [7]. When the effect on quality of life was included, the total rose to about €40,850. With no real treatment and costs this high, there is great need for trials that can show whether a new therapy works.

### Disease progression & prognosis
FSHD usually gets worse slowly over decades. About 20% of patients eventually need a wheelchair, but life expectancy is generally normal, as severe cardiac involvement or breathing failure is rare [8]. Progression also happens muscle by muscle rather than all at one time. In a one-year MRI study, the overall fat in the legs barely changed, but individual muscles changed drastically, and muscles that started at a medium fat level (40 to 50%) progressed the most rapid [9]. In that study, the tibialis anterior changed very little over one year, and TA strength dropped sharply once more than about 20% of the muscle was replaced by fat. Muscles that are bright on STIR MRI (a sign of inflammation or edema) are much more likely to get fattier: 13.9% of STIR-positive muscles had more fat after a year, compared to only 0.21% of STIR-negative muscles [10]. Meanwhile, functional tests like the 6-minute walk didn't change noticeably in a year. This is the core problem for our project: a short trial needs a measure that changes within months, and MRI can, but only in certain muscles, while function may lag behind.

### Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
FSHD is caused by a gene called DUX4 that is inside each D4Z4 repeat. DUX4 is normally switched off in adult muscle, but when the repeat array is shortened (FSHD1) or loses its methylation (FSHD2), the DNA opens. On the 4qA chromosome version, the DUX4 message gets a signal that makes it stable, so muscle cells start making DUX4 protein. DUX4 turns on genes that are normally only active in early embryos, along with immune-related genes, and this is toxic to muscle cells and leads to cell death [11]. At the tissue level, the lost muscle fibers are replaced by fat and scar tissue, along with chronic inflammation. Fat replacement is what the MRI fat fraction measures in our data set. In the study our dataset comes from, TA muscles that were STIR-positive had more fat and higher DUX4 activity, every TA over 20% fat was STIR-positive, and higher DUX4 activity was linked to lower TA strength and worse clinical severity [12]. At the whole muscle level, strength depends on both how much muscle there is and its quality, since fat infiltration lowers how much force each part of the muscle can produce. This explains why fat fraction should be related to strength.

### Clinical trials/next-gen therapies
Since DUX4 is the root cause of FSHD, most new therapies try to turn it off. One attempt, losmapimod, made it to a Phase 3 trial but didn’t beat placebo on its main measure, reachable workspace, and was stopped in 2024 [13]. Delpacibart braxlosiran (del-brax), owned by Novartis, uses an antibody attached to a small piece of RNA to silence DUX4 directly in muscle. In early trials, it cut DUX4-related genes by more than half with no serious side effects, and it is now in a 200-patient Phase 3 trial that measures muscle strength and walking speed [14]. EPI-321, made by Epicrispr Biotechnologies, takes a different approach: it’s a one-time IV infusion that uses epigenetic editing to keep DUX4 switched off without changing the DNA itself. Early Phase 1/2 results showed patients gained from 0.5 to 1.3 pounds of lean muscle on MRI after six months with no serious side effects so far, though only three patients had reached that point [15].
